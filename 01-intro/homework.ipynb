```python
df_q7 = df_asia[['vehicle_weight', 'model_year']]
df_q7.head()
```

```python
df_q7 = df_q7.iloc[[0,1,2,3,4,5,6]]
df_q7
```

```python
X = np.vstack([df_q7.vehicle_weight.values, df_q7.model_year.values])
X
```

```python
XTX = X.dot(X.T)
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
X.T
```

```python
w_t = (X.T).dot(XTX_inv)
w_t
```

```python
w = y.dot(w_t)
w
```
